# M5Atom S3 Camera MJPG viewer
Displays the MJPG stream from http://192.168.4.1, refreshing every second.

```python
from IPython.display import Image, display, clear_output
import time

STREAM_URL = 'http://192.168.4.1'
REFRESH_SECONDS = 1

while True:
    clear_output(wait=True)
    # Cache-bust so the browser fetches a fresh frame
    frame_url = f"{STREAM_URL}?t={time.time()}"
    display(Image(url=frame_url))
    time.sleep(REFRESH_SECONDS)
```
