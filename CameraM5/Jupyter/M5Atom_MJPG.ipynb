# M5Atom S3 Camera MJPG viewer
Displays the MJPG stream from http://192.168.4.1, refreshing every second.

```python
from IPython.display import display, clear_output
import time
import requests
import numpy as np
import cv2
import matplotlib.pyplot as plt

CAPTURE_URL = 'http://192.168.4.1/capture'
REFRESH_SECONDS = 0.1
session = requests.Session()

while True:
    try:
        resp = session.get(CAPTURE_URL, timeout=2, headers={'Connection': 'close'})
        if resp.status_code != 200:
            time.sleep(REFRESH_SECONDS)
            continue
        frame = cv2.imdecode(np.frombuffer(resp.content, np.uint8), cv2.IMREAD_GRAYSCALE)
        if frame is None:
            time.sleep(REFRESH_SECONDS)
            continue
        clear_output(wait=True)
        plt.imshow(frame, cmap='gray')
        plt.axis('off')
        display(plt.gcf())
        plt.close()
    except Exception:
        time.sleep(REFRESH_SECONDS)
```
