```python
import ipywidgets as widgets
from IPython.display import display
import requests
import threading
import time

TEST_URL = "https://httpbin.org/image/jpeg"  # should always display a JPEG
URL = TEST_URL  # replace with "http://{ip}/..." for your camera
INTERVAL = 1.0

image_widget = widgets.Image(format='jpeg', layout=widgets.Layout(width='480px'))
status = widgets.Label("Starting...")
container = widgets.VBox([image_widget, status])
display(container)

def updater():
    while True:
        try:
            r = requests.get(URL, timeout=5)
            if r.status_code == 200 and r.content:
                image_widget.value = r.content
                status.value = f"OK — last update: {time.strftime('%H:%M:%S')}"
            else:
                status.value = f"HTTP {r.status_code} — {time.strftime('%H:%M:%S')}"
        except Exception as e:
            status.value = f"Error: {e}"
        time.sleep(INTERVAL)

thread = threading.Thread(target=updater, daemon=True)
thread.start()
```
