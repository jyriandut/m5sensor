# Simulatsioonis asenda HMAC-SHA256 algoritm XOR'iga

see on tahtlikult nõrk, et simulatsioon oleks lahendatav mõistliku ajaga. Näita Jupyter notebook'is kuidas ülevõtmine töötab: tee kõigepealt muutuja secret key ja siis proovi seda samm-sammult ära arvata.

```python
import os
import itertools

secret_key = os.urandom(3) # Kasuatame 3 byte pikkust võtit
msg = b"HELLO"
# XOR-ime sõnumi oma secret keyga
# "Tag" = XOR with repeating key (toy, weak)
cipher = bytes([m ^ secret_key[i % len(secret_key)] for i, m in enumerate(msg)])
print(f"Cytpher {cipher}")

# brute-force key (3 bytes)
found = None
for k in itertools.product(range(256), repeat=3):
  key = bytes(k)
  guess = bytes([m ^ key[i % 3] for i, m in enumerate(msg)])
  if guess == cipher:
      found = key
      break

print("secret:", secret_key)
print("found:", found)
print("match:", found == secret_key)
```
