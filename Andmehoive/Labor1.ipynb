```python
PORT = "/dev/cu.usbserial-615A4E2C46"   # <- replace with your device
BAUD = 115200
```

```python
WINDOW_SECONDS = 60.0     # how much history to show
SAMPLE_HZ_GUESS = 10      # rough guess (used to size buffer)
OUTPUT = "pressure_log.csv"  # change or None

ser = serial.Serial(PORT, BAUD, timeout=1)
```

```python
with open(OUTPUT, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["timestamp", "raw_line"])
    print(f"Recording to {OUTPUT}... (press Ctrl+C to stop)")
    try:
        while True:
            line = ser.readline().decode("utf-8", errors="ignore").strip()
            if not line:
                continue
            writer.writerow([time.strftime("%Y-%m-%d %H:%M:%S"), line])
            print(line)
    except KeyboardInterrupt:
        print("\nStopped.")
```
